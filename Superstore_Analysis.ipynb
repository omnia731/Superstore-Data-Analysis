## 5.Object-Oriented Analytical Framework

A reusable `DataAnalyzer` class is created to organize preprocessing validation, feature engineering, analytical functions, visualizations, KPI generation, and export operations.

```python
class DataAnalyzer:

    def __init__(self, df):
        self.df = df

    ###############################################Data Preprocessing#############################################################
    @staticmethod
    def preprocess_data(df):

        clean_df = df.copy()

        # Handle missing values

        for col in clean_df.columns:

            if clean_df[col].isnull().sum() > 0:

                if col == "Postal Code":
                    clean_df[col] = clean_df[col].fillna(
                        clean_df[col].mode()[0]
                    )

                else:
                    clean_df[col] = clean_df[col].fillna(
                        clean_df[col].median()
                    )

        # Check and convert date columns

        date_columns = ["Order Date", "Ship Date"]

        for col in date_columns:

            if not pd.api.types.is_datetime64_any_dtype(clean_df[col]):
                clean_df[col] = pd.to_datetime(clean_df[col])

        # Remove duplicate records

        clean_df = clean_df.drop_duplicates()

        # Remove leading and trailing spaces

        text_columns = clean_df.select_dtypes(include="object").columns

        for col in text_columns:
            clean_df[col] = clean_df[col].str.strip()

        return clean_df
    
    
       ################################################ Feature Engineering #############################################################
    def add_profit_margin(self):

        try:
            if "Sales" not in self.df.columns:
                raise KeyError("Sales")

            if "Profit" not in self.df.columns:
                raise KeyError("Profit")

            if self.df["Sales"].eq(0).any():
                raise ZeroDivisionError("Sales contains zero values.")

            self.df["Profit Margin"] = (
                self.df["Profit"] / self.df["Sales"]
            ) * 100

        except ZeroDivisionError as e:
            print(f"Error: {e}")

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error while calculating Profit Margin: {e}")



    def add_shipping_duration(self):
        try:
            self.df["Shipping Duration"] = (
                self.df["Ship Date"] - self.df["Order Date"]
            ).dt.days

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error while calculating Shipping Duration: {e}")

    def add_sales_performance_category(self):

        try:
            self.df["Sales Performance Category"] = pd.qcut(
                self.df["Sales"],
                q=3,
                labels=["Low", "Medium", "High"]
            )

        except Exception as e:
            print(f"Error while creating Sales Performance Category: {e}")
            
         
       ############################################### Memory Optimization #############################################################
            
    def optimize_memory(self):

        try:
            before = self.df.memory_usage(deep=True).sum() / 1024**2

            for col in self.df.columns:

                if self.df[col].dtype == "int64":
                    self.df[col] = self.df[col].astype("int32")

                elif self.df[col].dtype == "float64":
                    self.df[col] = self.df[col].astype("float32")

                elif self.df[col].dtype == "object":

                    if self.df[col].nunique() / len(self.df[col]) < 0.5:
                        self.df[col] = self.df[col].astype("category")

            after = self.df.memory_usage(deep=True).sum() / 1024**2

            reduction = before - after
            percentage = (reduction / before) * 100

            print("Memory Optimization")
            print("-" * 40)
            print(f"Memory Before : {before:.2f} MB")
            print(f"Memory After  : {after:.2f} MB")
            print(f"Memory Reduced: {reduction:.2f} MB")
            print(f"Reduction     : {percentage:.2f}%")

        except Exception as e:
            print(f"Error optimizing memory usage: {e}")          
            
       ################################################ Modular Analytical Functions #############################################################
     
    def calculate_kpis(self):

        try:
            kpis = {
                    "Total Sales": self.df["Sales"].sum(),
                    "Total Profit": self.df["Profit"].sum(),
                    "Average Sales": self.df["Sales"].mean(),
                    "Average Profit": self.df["Profit"].mean(),
                    "Total Quantity": self.df["Quantity"].sum()
                }
            
            kpi_df = pd.DataFrame(
                    list(kpis.items()),
                    columns=["KPI", "Value"]
                )
            
            kpi_df["Value"] = kpi_df["Value"].apply(lambda x: f"{x:,.2f}")
            
            return kpi_df

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error while calculating KPIs: {e}")       
            
        
    def get_loss_making_products(self):

        try:
            product_profit = (
                self.df.groupby("Product Name", as_index=False)["Profit"]
                .sum()
                .sort_values("Profit")
                .reset_index(drop=True)
            )

            return product_profit[product_profit["Profit"] < 0]

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error getting loss-making products: {e}")
            
            
    def find_high_sales_low_profit(self):
        try:
            sales_threshold = self.df["Sales"].quantile(0.75)
            profit_threshold = self.df["Profit"].quantile(0.25)

            return self.df[
                (self.df["Sales"] >= sales_threshold) &
                (self.df["Profit"] <= profit_threshold)
            ]

        except KeyError as e:
            print(f"Error: Missing required column - {e}")
            
            
            
    def get_discount_impact(self):
        try:
            return (
                self.df.groupby("Discount")
                .agg(
                    Average_Sales=("Sales", "mean"),
                    Average_Profit=("Profit", "mean"),
                    Total_Profit=("Profit", "sum")
                )
                # .sort_index()
                .reset_index()
            )

        except KeyError as e:
            print(f"Error: Missing required column - {e}")
            
            
    def correlation_matrix(self, columns):

        try:
            return self.df[columns].corr()

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error calculating correlation matrix: {e}")
            
                     
       ############################################### Visualizations #############################################################

    def plot_distribution(self, column, save_path=None):

        try:
            plt.figure(figsize=(10, 6))

            sns.histplot(
                data=self.df,
                x=column,
                bins=30,
                kde=True
            )

            plt.title(f"{column} Distribution")
            plt.xlabel(column)
            plt.ylabel("Frequency")

            plt.tight_layout()

            if save_path is not None:
                plt.savefig(save_path, dpi=300, bbox_inches="tight")

            plt.show()
            plt.close()

        except KeyError as e:
            print(f"Error: Column '{e.args[0]}' not found in the dataset.")

        except Exception as e:
            print(f"Error creating distribution plot: {e}")
            
            
    def plot_total_by_group(self, group_column, value_column, save_path=None):

        try:
            grouped_data = (
                self.df.groupby(group_column)[value_column]
                .sum()
            )

            plt.figure(figsize=(8, 5))

            grouped_data.plot(kind="bar")

            plt.title(f"Total {value_column} by {group_column}")
            plt.xlabel(group_column)
            plt.ylabel(f"Total {value_column}")

            plt.xticks(rotation=0)

            plt.tight_layout()

            if save_path is not None:
                plt.savefig(
                    save_path,
                    dpi=300,
                    bbox_inches="tight"
                )

            plt.show()
            plt.close()

        except KeyError as e:
            print(
                f"Error: Column '{e.args[0]}' not found in the dataset."
            )

        except Exception as e:
            print(
                f"Error creating grouped bar chart: {e}"
            )
            
            
    def plot_time_trend(self,date_column,value_column,freq="ME",save_path=None):

        try:
            trend_data = (
                self.df
                .groupby(
                    pd.Grouper(
                        key=date_column,
                        freq=freq
                    )
                )[value_column]
                .sum()
            )

            plt.figure(figsize=(12, 6))

            plt.plot(
                trend_data.index,
                trend_data.values,
                marker="o"
            )

            plt.title(f"{value_column} Trend Over Time")
            plt.xlabel("Month")
            plt.ylabel(f"Total {value_column}")

            plt.xticks(rotation=45)

            plt.tight_layout()

            if save_path is not None:
                plt.savefig(
                    save_path,
                    dpi=300,
                    bbox_inches="tight"
                )

            plt.show()
            plt.close()

        except KeyError as e:
            print(
                f"Error: Column '{e.args[0]}' not found in the dataset."
            )

        except Exception as e:
            print(
                f"Error creating time trend plot: {e}"
            )
            
            

    @staticmethod        
    def plot_scatter(data, x_column, y_column,hue_column=None, save_path=None):

        try:
            plt.figure(figsize=(10, 6))

            sns.scatterplot(
                data=data,
                x=x_column,
                y=y_column,
                hue=hue_column
            )
            
            plt.axhline(0, linestyle="--")
            plt.title(f"{y_column} vs {x_column}")
            plt.xlabel(x_column)
            plt.ylabel(y_column)

            plt.tight_layout()

            if save_path:
                plt.savefig(save_path, dpi=300, bbox_inches="tight")

            plt.show()

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error creating scatter plot: {e}")
            
            
    def plot_heatmap(self, row_column, column_column, value_column, save_path=None):

        try:
            heatmap_data = self.df.pivot_table(
                index=row_column,
                columns=column_column,
                values=value_column,
                aggfunc="mean"
            )

            plt.figure(figsize=(10, 6))

            sns.heatmap(
                heatmap_data,
                annot=True,
                fmt=".2f"
            )

            plt.title(f"Average {value_column} by {row_column} and {column_column}")
            plt.xlabel(column_column)
            plt.ylabel(row_column)

            plt.tight_layout()

            if save_path:
                plt.savefig(
                    save_path,
                    dpi=300,
                    bbox_inches="tight"
                )

            plt.show()

        except KeyError as e:
            print(f"Error: Missing required column - {e}")

        except Exception as e:
            print(f"Error creating heatmap: {e}")        
            
    

       ################################################ Generating Report #############################################################
   
    
    def generate_report(self):

        try:
            kpis = self.calculate_kpis()

            loss_products = self.get_loss_making_products()

            unprofitable_sales = self.find_high_sales_low_profit()

            discount_impact = self.get_discount_impact()

            print("\n" + "=" * 70)
            print("                    KPI SUMMARY")
            print("=" * 70)

            for _, row in kpis.iterrows():
                print(f"{row['KPI']:<25}: {row['Value']}")

            print("\n" + "=" * 70)
            print("                  BUSINESS OVERVIEW")
            print("=" * 70)

            print(
                f"{'Loss-Making Products':<25}: "
                f"{len(loss_products)}"
            )

            print(
                f"{'Unprofitable Sales':<25}: "
                f"{len(unprofitable_sales)}"
            )

            print("\n" + "=" * 70)
            print("                    DISCOUNT IMPACT")
            print("=" * 70)

            print(
                discount_impact.to_string(
                    index=False,
                    float_format=lambda x: f"{x:,.2f}"
                )
            )

            print("=" * 70)

            return {
                "KPIs": kpis,
                "Loss-Making Products": loss_products,
                "Unprofitable Sales": unprofitable_sales,
                "Discount Impact": discount_impact
            }

        except Exception as e:
            print(f"Error generating report: {e}")
            return None
            
            
       ################################################ Export cleaned data #############################################################
            
    def export_cleaned_data(self, file_path="cleaned_superstore.csv"):

        try:
            self.df.to_csv(file_path, index=False)

            print(f"Cleaned dataset exported successfully to: {file_path}")

        except Exception as e:
            print(f"Error exporting cleaned dataset: {e}")
```
